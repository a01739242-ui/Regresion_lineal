```python
# Seleccionamos únicamente las variables de la regresión

variables_regresion = df[['Vacantes','Auditoría Salesforce', 'Capacitación de Marca', 'Indice de clima laboral', 'Productividad de APVS','Cumplimiento de plantilla']]
variables_regresion = variables_regresion.fillna(0)
```

```python
from turtle import color
sns.scatterplot(x='Productividad de APVS', y='Cumplimiento de plantilla', color="red", data=df)
```

```python
#Declaramos las variables dependientes e independientes para la regresión lineal
Vars_Indep= variables_regresion[['Productividad de APVS']]
Var_Dep= variables_regresion['Cumplimiento de plantilla']
```

```python
%pip install scikit-learn
```

```python
#Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model= LinearRegression()
```

```python
#Ajustamos el modelo con las variables antes declaradas
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
#Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

```python
#Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep,Var_Dep)
```

```python
y_pred= model.predict(X=variables_regresion[['Productividad de APVS']])
y_pred
```

```python
#Insertamos la columna de predicciones en el DataFrame
variables_regresion.insert(0, 'Predicciones', y_pred)
variables_regresion
```

```python
sns.scatterplot(x='Productividad de APVS', y='Cumplimiento de plantilla', color="blue", data=variables_regresion)
sns.scatterplot(x='Productividad de APVS', y='Predicciones', color="red", data=variables_regresion)
sns.lineplot(x='Productividad de APVS', y='Predicciones', color="red", data=variables_regresion)
```

```python
coef_Deter=model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
#Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl=np.sqrt(coef_Deter)
coef_Correl
```

```python
#Graficamos todas las dispersiones entre todas las variables
sns.pairplot(variables_regresion)
```

```python
#Encontramos todas las correlaciones entre las variables 
Corr_Factors=variables_regresion.corr()
Corr_Factors
```

```python
#Encontramos el valor absoluto de todas las correlaciones entre las variables 
Corr_Factors1= abs(Corr_Factors)
Corr_Factors1
```

```python
#Ajustamos el mapa de calor de los coeficientes de correlación 
Heat_Map= sns.heatmap(Corr_Factors1, cmap = 'Reds', annot=True, fmt=".2f")  
Heat_Map
```
